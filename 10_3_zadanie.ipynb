```python
import numpy as np
import pandas as pd
from scipy import stats
from sklearn.preprocessing import StandardScaler, MinMaxScaler

data_100 = np.random.normal(loc=3, scale=1, size=100)
data_10000 = np.random.normal(loc=3, scale=1, size=10000)
```

```python
# Zbiór 100

median_100 = np.median(data_100)
mode_100 = stats.mode(data_100)
max_100 = np.max(data_100)
min_100 = np.min(data_100)
q1_100 = np.quantile(data_100, 0.25)
q3_100 = np.quantile(data_100, 0.75)

print(f'Mediana {median_100}')
print(f'Moda {mode_100}')
print(f'Max {max_100}')
print(f'Min {min_100}')
print(f'Kwartyl Q1 {q1_100}')
print(f'Kwartyl Q3 {q3_100}')

```

```python
# Zbiór 10 000

median_10000 = np.median(data_10000)
mode_10000 = stats.mode(data_10000)
max_10000 = np.max(data_10000)
min_10000 = np.min(data_10000)
q1_10000 = np.quantile(data_10000, 0.25)
q3_10000 = np.quantile(data_10000, 0.75)

print(f'Mediana {median_10000}')
print(f'Moda {mode_10000}')
print(f'Max {max_10000}')
print(f'Min {min_10000}')
print(f'Kwartyl Q1 {q1_10000}')
print(f'Kwartyl Q3 {q3_10000}')
```
