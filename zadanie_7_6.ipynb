```python
import numpy as np
import matplotlib.pyplot as plt
import numpy_financial as npf

home_price = 120000
rate = 0.05
years = 5

home_end_value = npf.fv(rate=rate, nper=years, pmt=0, pv=home_price)

# parametry lokaty
month_rate = 0.12 / 12
month_number = 5 * 12

# miesięczna wpłata
monthy_payment = -npf.pmt(rate=month_rate,
                nper=month_number,
                pv=0,
                fv=home_end_value)

print(f"Wartość mieszkania po {years} latach: {home_end_value:.2f} zł")
print(f"Miesięczna wpłata: {monthy_payment:.2f} zł")
```

Wartość mieszkania po 5 latach: -153153.79 zł
Miesięczna wpłata: -1875.28 zł

```python
# Dane

months = 60
monthly_rate = 0.12 / 12  # 1% miesięcznie

# Oś czasu
t = np.arange(months + 1)

# Cena mieszkania - liniowy wzrost
prices = np.linspace(home_price, home_end_value, months + 1)

# Lokata
savings = []
capital = 0

for m in range(months + 1):
    savings.append(capital)
    capital = capital * (1 + monthly_rate) + monthy_payment

# Wykres
plt.figure(figsize=(10, 6))

plt.plot(t, prices, label="Cena mieszkania", linewidth=2)
plt.plot(t, savings, label="Wartość lokaty", linewidth=2)

plt.title("Cena mieszkania vs wartość lokaty (60 miesięcy)")
plt.xlabel("Miesiące")
plt.ylabel("PLN")
plt.grid(True)
plt.legend()

plt.show()
```
