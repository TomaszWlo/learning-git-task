```python
import pandas as pd

movies = pd.read_csv('tmdb_movies.csv')
kwartyl = movies['vote_count'].quantile(0.75)

top_10 = movies[movies['vote_count'] > kwartyl].sort_values('vote_average', ascending=False).head(10)

print(top_10[["title", "vote_average", "vote_count"]])
```

```python
import matplotlib.pyplot as plt
import pandas as pd

movies = pd.read_csv('tmdb_movies.csv')

movies["release_date"] = pd.to_datetime(movies['release_date'])
movies['year'] =  movies["release_date"].dt.year

year_range = movies[(movies["year"] >= 2010) & (movies["year"] <= 2016)]
movie_stats = year_range.groupby('year')[["revenue", "budget"]].mean()

fig, axes = plt.subplots(figsize=(10, 6))
movie_stats["revenue"].plot(
    kind="bar",
    ax=axes,
    color="skyblue",
    label="Średni przychód"
)

movie_stats["budget"].plot(
    kind="line",
    ax=axes,
    color="red",
    marker="o",
    label="Średni budżet"
)

axes.set_title("Średni przychód i budżet filmów w latach 2010–2016")
axes.set_xlabel("Rok")
axes.set_ylabel("Kwota (USD)")
axes.ticklabel_format(style='plain', axis='y')
axes.legend(loc="upper left", bbox_to_anchor=(1.02, 1))
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


```

```python
import pandas as pd

movies = pd.read_csv("tmdb_movies.csv")
genres = pd.read_csv("tmdb_genres.csv")

table = pd.merge(movies,genres,on='genre_id', how='left')

genre_number = table.groupby("genres")["genre_id"].value_counts()

genre_number.sort_values(ascending=False)
```
