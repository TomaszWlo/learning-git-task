```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats

from sklearn.preprocessing import StandardScaler, MinMaxScaler

titanic = pd.read_csv('titanic.csv')

fare_values = titanic.loc[titanic['Fare'].notnull(), 'Fare'].values
print(fare_values.shape)

plt.hist(fare_values)

plt.title("Histogram cen biletów na Titanicu")
plt.xlabel("Cena biletu (Fare)")
plt.ylabel("Liczba pasażerów")

plt.show()
```

Z histogramu wynika ze najwiecej pasazerow bo ponad 700 wykupilo bilety do kwoty 50 $

```python
mean_fare = np.round(np.mean(fare_values))
median_fare = np.median(fare_values)
mode_fare = stats.mode(fare_values)

print(mean_fare)
print(median_fare)
print(mode_fare)
```

```python
Średnia wartosc bieltow wynosi 32$, mediana wynosi 14,5$. Najczesciej kupowanym biletem byl bilet o wartosci 8.05$ i bylo ich 43.
```

```python
plt.hist(fare_values)
plt.axvline(x = mean_fare, color='red', label = 'Średnia')
plt.axvline(x = median_fare, color='orange', label = 'Mediana')
plt.axvline(x = mode_fare[0], color='black', label = 'Moda')
plt.legend(loc='upper right')

plt.title("Histogram cen biletów na Titanicu")
plt.xlabel("Cena biletu (Fare)")
plt.ylabel("Liczba pasażerów")

plt.show()
```

```python
q0 = np.quantile(fare_values, 0.0)
q1 = np.quantile(fare_values, 0.25)
q2 = np.quantile(fare_values, 0.5)
q3 = np.quantile(fare_values, 0.75)
q4 = np.quantile(fare_values, 1.0)

print(f'Kwartyl 0 - {q0}')
print(f'Kwartyl 1 - {q1}')
print(f'Kwartyl 2 - {q2}')
print(f'Kwartyl 3 - {q3}')
print(f'Kwartyl 4 - {q4}')

```

```python
Pierwszy kwartyl wartosci biletow wynosi prawie 8$, drugi kwartyl ktry jest mediana wynosi 14,5$. Trzeci kwartyl czyli 75% calosci wynosi 31$ a najdrozszy bilet kosztowal ponad 512$
```
