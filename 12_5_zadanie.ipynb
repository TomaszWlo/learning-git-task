```python
from sklearn.datasets import load_digits
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split

digits = load_digits()
wine = load_wine()
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier

knn_clf = KNeighborsClassifier()
dt_clf = DecisionTreeClassifier()

X = digits.data
y = digits.target

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=18)

knn_clf.fit(X_train, y_train)
dt_clf.fit(X_train, y_train)

knn_train_score = knn_clf.score(X_train, y_train)
knn_test_score = knn_clf.score(X_test, y_test)

dt_train_score = dt_clf.score(X_train, y_train)
dt_test_score = dt_clf.score(X_test, y_test)

print('Digits')

print()
print("KNN")
print("Train score:", knn_train_score)
print("Test score:", knn_test_score)

print()

print("Decision Tree")
print("Train score:", dt_train_score)
print("Test score:", dt_test_score)
```

1. Do wykonania zadania uzylem generatora danych Digits
2. Uzyte klasyfiaktory to KNeighborsClassifier oraz
DecisionTreeClassifier
3. Dane zostaly podzielone na 70% danych treningowych oraz 30% danych testowych
4. Klasyfikator KNeighborsClassifier dziala bardzo dobrze różnica między wynikiem testowym a treningowym jest mała.
5. Klasyfiaktor DecisionTreeClassifier poradzil sobe gorzej roznica miedzy wynikierm testowym a treningowym jest widoczna. Uznałbym to za Overfitting









```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier

knn_wine = KNeighborsClassifier()

X = wine.data
y = wine.target

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=20)

knn_wine = KNeighborsClassifier()
dt_wine = DecisionTreeClassifier()

knn_wine.fit(X_train, y_train)
dt_wine.fit(X_train, y_train)

knn_wine_train_score = knn_wine.score(X_train, y_train)
knn_wine_test_score = knn_wine.score(X_test, y_test)

dt_wine_train_score = dt_wine.score(X_train, y_train)
dt_wine_test_score = dt_wine.score(X_test, y_test)


print("WINE")

print()
print("KNN:")
print("Train score:", knn_wine_train_score)
print("Test score:", knn_wine_test_score)

print()
print("Decision Tree:")
print("Train score:", dt_wine_train_score)
print("Test score:", dt_wine_test_score)
```

1. Do wykonania zadania uzylem generatora danych Wine
2. Uzyte klasyfiaktory to KNeighborsClassifier oraz
DecisionTreeClassifier
3. Dane zostaly podzielone na 80% danych treningowych oraz 20% danych testowych
4. Klasyfikator KNeighborsClassifier dziala nie najlepiej różnica miedzy danymi treningowymi a testowymi wynosi 10,2 punktu. Sam model nie radzi sobie najllepiej. Mozna wywnioskować niezbyt silny overfitting.
5. Klasyfiaktor DecisionTreeClassifier poradzil sobe bardzo dobrze. Wynik testowy jest bliski wynikowi treningowemu.
