```python
import pandas as pd

album = pd.read_html('https://www.officialcharts.com/chart-news/the-best-selling-albums-of-all-time-on-the-official-uk-chart__15551/', header=0)
df = album[0]

df.rename(columns={'TITLE':'TYTUŁ', "ARTIST":'ARTYSTA', 'YEAR':'ROK', 'HIGH POSN':'MAX POZ'}, inplace=True)
print(df.columns)

df.columns = df.columns.str.capitalize()
print(df.columns)

df.drop('Max poz', axis=1, inplace=True)
print(df.columns)

```

```python
df['Rok'].value_counts().head(7).sort_values(ascending=False)
```

```python
print(df['Rok'].max())
```
